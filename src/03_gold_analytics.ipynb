```python
gold_sales_data=spark.read.table("workspace.ibm.capstone_gold_sales")
gold_sales_data.show(5)
```

```python
from pyspark.sql.functions import *
total_orders=gold_sales_data.count()
sales_count=gold_sales_data.groupBy(col("order_status")).agg(count("order_id").alias("sales_count"))
sales_summary=gold_sales_data.filter(col("order_status")=="Completed").agg(
    sum(col("quantity")).alias("units_sold"),
    sum(col("gross_amount")).cast("decimal(18,2)").alias("gross_sales"),
    sum(col("discount_amount")).cast("decimal(18,2)").alias("discount_amounts"),
    sum(col("net_amount")).cast("decimal(18,2)").alias("net_sales"),
    (
        sum(col("gross_amount")) / sum(col("quantity"))
    ).cast("decimal(18,2)").alias("gross_aov"),
    (
        sum(col("net_amount")) / sum(col("quantity"))
    ).cast("decimal(18,2)").alias("net_aov")
    
)

print("Total orders: ",total_orders)
sales_count.show()
sales_summary.show()

```
