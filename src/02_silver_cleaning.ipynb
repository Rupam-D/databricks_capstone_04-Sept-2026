```python
silver_sales_data=spark.read.table("workspace.ibm.capstone_silver_sales")

silver_sales_data.printSchema()
```

```python
##Tranformation

from pyspark.sql.functions import *

string_type_column=["order_id","customer_id","customer_name","city","state","product_id","product_name","category","payment_method","order_status","quality_flag"]

silver_sales_data=silver_sales_data.select(
    *[
        trim(col(c)).alias(c) if(c in string_type_column) else col(c)
        for c in silver_sales_data.columns
    ]
).withColumn("order_date",to_date(col("order_date")))\
.withColumn("quantity",col("quantity").cast("int"))\
.withColumn("unit_price", col("unit_price").cast("double"))\
.withColumn("discount_pct", col("discount_pct").cast("double"))\
.withColumn("gross_amount", col("gross_amount").cast("double"))\
.withColumn("discount_amount",col("discount_amount").cast("double"))\
.withColumn("net_amount",col("net_amount").cast("double"))\
.withColumn("year",year(col("order_date")))\
.withColumn("month",month(col("order_date")))\
.withColumn("month_name",date_format(col("order_date"),"MMMM"))


silver_sales_data.printSchema()

```
